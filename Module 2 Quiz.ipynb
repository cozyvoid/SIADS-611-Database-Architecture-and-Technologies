# Module 2 (Week 2) Quiz

---

###  1. Which came first in PostgreSQL?

- JSON

    PostgreSQL added the JSON data type before JSONB. JSON arrived in PostgreSQL 9.2; JSONB came later in 9.4.

### 2. What index type is typically used for JSONB in PostgreSQL?

- GIN

    GIN (Generalized Inverted Index) is commonly used with JSONB because it efficiently indexes keys and values contained inside JSON documents.

### 3. Which of the following is closest to JSON conceptually?

- CSV

	Conceptually, JSON is closest to CSV among these choices because both are text-based formats used to represent and exchange structured data.

### 4. Which of the following spurred the adoption of JSON?

- Moving web application logic into the browser

    As more application logic moved into browsers with JavaScript/AJAX, applications needed a lightweight way to exchange data between browsers and servers. JSON fit this extremely well.

### 5. What programming language served as the inspiration for JSON?

- JavaScript

    JSON stands for JavaScript Object Notation and its syntax was inspired by JavaScript object literals.

### 6. What was a popular wire protocol format before JSON?

- XML

    XML was one of the dominant formats for exchanging structured data over the web before JSON became popular.

### 7. Which of the following is an equivalent to the concept of deserialization when it comes to wire protocols?

- unmarshalling

    Deserialization converts transmitted/stored data back into usable objects/data structures. Unmarshalling is an equivalent term frequently used when discussing wire protocols.
