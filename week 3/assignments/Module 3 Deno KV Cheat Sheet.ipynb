## 2. Deno Book Data Model

Reuse the same `kvadmin.py` and `hidden.py`. If KVAdmin is still at
`Enter command:`, continue there. If you see the shell prompt ending
in `$`, start KVAdmin again:

```bash
cd /voc/work
python kvadmin.py
```

Enter one row at a time. Paste the command at `Enter command:` and press
**Enter**. Paste its value at `Enter json` and press **Enter twice**.
Wait for `"ok": true` before entering the next command.

| Entry | Command | JSON value |
|---|---|---|
| ISBN record | `set /books/isbn/1730` | `{"author":"Vittore","title":"Tutti","isbn":"1730"}` |
| Author index | `set /books/author/Vittore/1730` | `{}` |
| Title index | `set /books/title/Tutti/1730` | `{}` |

The ISBN key stores the full book. Each secondary index stores an empty
object because the ISBN is already part of its key.

Click **Check** on the **Deno Book Data Model** assignment page. Once you
see the pass message, type `quit` at `Enter command:` and press **Enter**.

## Troubleshooting

| What you see | What to do |
|---|---|
| `can't open file '/voc/kvadmin.py'` | Run `cd /voc/work`, then `python kvadmin.py`. |
| `Invalid command` after pasting JSON | Enter the `set` command first. Paste the JSON when `Enter json` appears. |
| KVAdmin is still waiting after you enter JSON | Press **Enter** on an empty line to submit it. |
| KVAdmin shows `"ok": true`, but the grader shows `"value": null` | Compare the key in the grader’s `/kv/get/` URL with the key you entered after `set`. |
| The grader says a secondary key needs an empty object | Run `set` again for that author or title key and enter `{}` to overwrite its value. |
| The final grader step says `Missing or invalid token` | That step deliberately uses a bad token. Check for the overall pass message below it. |
