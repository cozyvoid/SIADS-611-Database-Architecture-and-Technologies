# Module 3 (Week 3) Quiz

---

### 1. What project is similar to and came before Deno?

    > Node

	Node.js is the earlier JavaScript runtime created by Ryan Dahl, who later created Deno.

### 2. What language does Deno use?

    > JavaScript

    Deno runs JavaScript and also supports TypeScript.

### 3. When you put your application into Deno Deploy, where does your code execute?

    > Various Deno servers around the world depending on your load and where your application is being used

    The course describes Deno Deploy running application code across distributed locations.

### 4. What is Deno KV Watch?

    > The ability to register an interest in a Deno document and be notified if it changes.

### 5. Which of the following key types is available in Deno KV?

    > Hierarchical Logical Key

    kv.watch() lets you subscribe to updates for specific keys.

### 6. Which of the following is *not* a peer product to Deno?

    > Amazon Elasticache

    ElastiCache is a managed caching service. The other choices offer application hosting or deployment platforms.

### 7. Which of the following is a feature of a free CloudFlare account?

    > Distributed Denial of Service Mitigation

	Cloudflare includes DDoS protection on its free plan.

### 8. Which of the following database technologies is used in the enterprise-scale Deno Deploy KV?

    > Foundation DB

    FoundationDB is the database technology behind the managed Deno Deploy KV described in the course.

### 9. Which of the following SQL Concepts works in Deno KV?

    > Logical Key indexing and lookup

    Deno KV supports direct lookup by key and listing by key prefix.

### 10. Which of the following Python data structures is closest to Deno KV?

    > A dictionary with tuples as keys

    A Python key like ("users", "alice") is analogous to the Deno KV key ["users", "alice"].

### 11. In the lecture example of eventual consistency using multi-server distributed B-Trees, what key value is inserted?

    > 04

    Slides 26–30 show set /04. One node initially knows about the new key, and eventually all nodes know where to find it.

### 12. Which of the following statements best characterizes the state of a multi-server B-Tree store between the time an insert starts on one server and is fully completed across all servers?

    > The record may or may not see the record depending on which server is asking the B-Tree

    During replication, one server may already return the new record while another still returns the older state.

### 13. What does "var" mean when applied to a JavaScript variable?

    > The variable may be changed in later code

    A variable declared with var can be reassigned.

### 14. Which language was used as an inspiration for the overall syntax of JavaScript?

    > C

    JavaScript uses C-style syntax, including braces for blocks.

### 15. What is the equivalent data structure to a Python list in JavaScript?

    > Array

    Like a Python list, a JavaScript array is an ordered collection with zero-based indexing.

### 16. What is the closest equivalent to the print() feature of Python in JavaScript?

    > console.log()

    It displays values in the console.

### 17. In what order were the following languages invented?

    > C / PYTHON / JavaScript / Ruby

    This is the intended choice. The wording “invented” is imprecise because Ruby development began before JavaScript, and both became public in 1995.

### 18. In the kvadmin.py application, what command shows example JSON documents?

    > samples

	Slide 40 shows samples entered at the kvadmin.py prompt, followed by two example JSON documents.

### 19. In Deno KV, what happens is you store a JSON document with a key of "hello" in one "set" request and store a document with a key of "world" in a second "set" request and then you execute a "get" to retrieve the document.

    > The document only has the "world" key

    Assuming both set calls use the same Deno KV key, the second call replaces the entire stored JSON value.

### 20. In the data modelling lecture, how do we emulate a secondary or "foreign key"?

    > We include the logical key as part of the secondary index logical key

    A key like ["books_by_author", author, bookId] lets you find all books by an author and identify each matching book.
