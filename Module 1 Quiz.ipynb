# Module 1 Quiz

---

1. Which of the following SCL features is most affected by the addition of an index on a table?

- `WHERE` clauses

2. Why does PostgreSQL store rows in blocvks with some space unused on disk?

- To allow for small changes or insertions of rows to be done without rewriting the entire file.

3. What is the typical block size used by PostgreSQL?

- 8K

4. What is a problem caused by a block size that is too small?

- Fragmentation

5. What is the advantage of a HASH index over a B-tree index in PostgreSQL?

- The B-tree must store the entire key value.

6. What kinds of key patterns work well with a BRIN index in PostgreSQL?

- When the rows in a block have key values in a nice tight range.

7. Which of the following is a PostgreSCL "inverted" index?

- GIN

8. Which of the following describes "stop word" in natural language indexing?

- A word  that contributes no meaning to a query

9. Which of the following is FALSE about Google's search?

- Google search used a single very large server for each section? WRONG

10. What does the PostgreSQL `unnest()` function do?

- It takes an array and expands it into a list of rows, one of each element in the array.

11. What Python function is most like the PostgreSQL `string_to_array()` function?

- `split()`

12. What does it mean when you add `DISTINCT` to a `SELECT` statement?

- All of the returned rows are unique.

13. In PostgreSQL 11, what is the operator class used to build an index to speed up queries that use the `<@` operator?

- `array_ops`

14. Why do we reduce words to their stems when building an inverted index for natural language documents?

- To map multiple words with equivalent meaning to a single stem

15. Which of the following is an advantage of the GiST over the GIN index in PostgreSQL?

- Fast inserts and updates

16. Which came first, the tsquery or the tsvector?

- tsvector

17. What function do you apply to a column to build a language full-text index on the column?

- tsvector

18. What PostgreSQL operator checks to see if a tsquery matches a tsvector?

- `@@`
